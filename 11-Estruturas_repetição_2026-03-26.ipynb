## Instancie um objeto da classe lista contendo 5.000 valores inteiros iniciando em 0. 
Quantos são pares e quantos são impares?

## 01 – Descreva o racional da sua resposta:

R.: Para criar a lista usa-se o método list() com um range de 5000 valores.
Depois é preciso uma condicional com dois contadores que devem ser zerados no início: pares e impares
A condicional inicia com um 'if' que avalia a cada valor da lista se o resto da divisão por 2 é igual a 0 (configurando um valor par).
Então, se for true a condição, soma-se um ao contador 'pares'
Se não, é aplicado o else no qual soma-se 1 ao contador 'impares'

ao final, após percorrer toda a lista e fora do loop, são retornados os valores finais dos contadores.

## 02 – Desenhe o fluxograma da resposta:

## 03 – Escreva o programa em Python:

```python
lista_5mil = list(range(5000))

pares=0
impares=0
for i in lista_5mil:
    if i % 2 == 0:
        pares+=1
    else:
        impares+=1

print(f'número de valores pares: {pares}')
print(f'número de valores impares: {impares}')
```

## 04 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
lista_5mil = list(range(5000))

pares = 0
impares = 0

i = 0 

while i < len(lista_5mil):
    if lista_5mil[i] % 2 == 0:
        pares += 1
    else:
        impares += 1
    i += 1  
    
print(f'número de valores pares: {pares}')
print(f'número de valores impares: {impares}')
```

## Importe o .csv exportado na última aula e instancie um objeto da classe DataFrame.

Dos números de letras nos genes, quais são pares e quantos são impares?

## 05 – Descreva o racional da sua resposta:

R.: Primeiro, deve-se importar a biblioteca pandas as pd neste notebook e importar com o comando indicado adicionando o parâmetro index_col="Unnamed: 0" para que os index sejam os nomes dos genes. As colunas permanecem como o original.

Para a pergunta, instancia-se uma lista vazia (genes_pares) e um contador (impares) igual a 0. Depois, entra a estrutura de repetição:
Nela, o for deve indicar que serão percorridos todos os itens (.items()) de uma series pandas criada a partir da coluna de número de caracteres (df_4_8['num_caract']) e considerar as duas posições de cada item, isto é a chave (index-genes) e o valor correspondente (values-contagem).

Nessa repetição cria-se uma condicional que vai analisar cada valor de contagem pelo resto da divisão por 2 e se o resultado for igual a zero, a chave equivalente a essa contagem é adicionada à lista criada. Caso o resultado seja diferente de zero, ou seja, false é executado o comando no else, o qual soma 1 ao contador de impares

Ao final, após todos os itens da series pandas, são retornados o valor do contador e a lista final.

## 06 – Desenhe o fluxograma da resposta:

## 07 – Escreva o programa em Python:

```python
import pandas as pd
df_4_8=pd.read_csv('df_4_8_7lista6.csv', index_col="Unnamed: 0")


genes_pares=[]
impares=0
for index, values in df_4_8["num_caract"].items():
    if values % 2 == 0:
        genes_pares.append(index)  
    else:
        impares += 1

print(f'Genes com número par de caracteres: {genes_pares}')
print(f'Número de genes com número ímpar de caracteres: {impares}')
```

## 08 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
genes_pares = []
impares = 0

itens = list(df_4_8["num_caract"].items()) 

i = 0

while i < len(itens):
    index, values = itens[i]
    
    if values % 2 == 0:
        genes_pares.append(index)
    else:
        impares += 1
    
    i += 1 

print(f'Genes com número par de caracteres: {genes_pares}')
print(f'Número de genes com número ímpar de caracteres: {impares}')
```
