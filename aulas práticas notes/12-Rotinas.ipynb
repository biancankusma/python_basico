```python
def contar_inteiros(obj:list|tuple)->int: #recebe o parametro obj, que pode ser uma lista ou tupla
    contador=0
    for x in range(len(obj)): #percorre a lista usando o indice x com base no numero de elementos da lista
        if type(obj[x]) == int:
            contador+=1
    return contador

#poderia por um valor default no parametro, como por exemplo: def contar_inteiros(obj:list|tuple=['a',1,True,'b',0,False])->int: 
#esse valor é usado se nao for fornecido um valor pro parametro na hora de usá-la
```

```python
#se quiser mais de um parametro, é só colocar uma virgula entre eles, como por exemplo: def contar_inteiros(obj:list|tuple, outro_parametro:int)->int:
#e na hora de usar a função, tem que passar os dois parametros, como por exemplo: contar_inteiros(['a',1,True,'b',0,False], 10)
#também dá pra criar um dicionário dos parametros que quero usar
#e depois no uso da função colocar função(**dicionário), que vai passar os parametros do dicionário pra função
```
