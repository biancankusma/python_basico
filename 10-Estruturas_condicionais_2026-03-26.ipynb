## 01 – Dados as operações abaixo, em Python, qual o resultado da operação lógica: 
p and q and r and s and t?
<br>
<br>Imprima os objetos e o resultado usando uma estrutura condicional.

<br>p = 2.0 + 3.0 / 5.0 == 10
<br>q = (p + 10.2) - 5.0 * 2.0 <= 2.9
<br>r = 10 <= p / q + 33.8 < 100
<br>s = 5 != p / (q + r)
<br>t = 33 == s * 2

```python
p = 2.0 + 3.0 / 5.0 == 10
q = (p + 10.2) - 5.0 * 2.0 <= 2.9
r = 10 <= p / q + 33.8 < 100
s = 5 != p / (q + r)
t = 33 == s * 2


if (p and q and r and s and t) == False:
    print('A operação lógica é falsa')
    print(f'p = {p}')
    print(f'q = {q}')
    print(f'r = {r}')
    print(f's = {s}')
    print(f't = {t}')
else:
    print('A operação lógica é verdadeira')
    print(f'p = {p}')
    print(f'q = {q}')
    print(f'r = {r}')
    print(f's = {s}')
    print(f't = {t}')
```

## Instancie um objeto da classe array numpy, contendo 5 valores quantitativos randômicos normalmente distribuídos, com média 10 e desvio padrão 3.

## 02 – Descreva o racional da sua resposta:

R.: Primeiro, para criar esse objeto é preciso importar a biblioteca numpy: import numpy as np

Depois, o método para obter um array de distribuição normal é: np.random.normal()

Dentro dos parenteses, vão os parametros separados por virgula: media, desvio padrao, numero de valores

Para instanciação do objeto: array_rand=np.random.normal(10,3,5)

Segundo, para responder a pergunta, pode ser usada uma estrutura de repetição que vai partir de um contador em zero e uma lista vazia 

Ao longo de um range definido pelo numero de elementos no array, vai verificar uma condição que se dá pelo if.

Se o valor observado for maior que 10, será acrescentado esse valor à lista e adicionado 1 ao contador. 

Para adicionar valores à lista pode ser usado o método append() levando em conta que são valores float e a indexação do loop a cada volta

Então, a repetição se dá sucessivamente até o final, quando vai devolver fora do loop a contagem de valores maiores que 10 e a lista com estes valores.

## 04 – Escreva o programa em Python:

```python
import numpy as np
array_rand=np.random.normal(10,3,5)
print(array_rand)

contador=0
lista=[]
for i in range(len(array_rand)):
    if array_rand[i] > 10:
        contador+=1
        lista.append(float(array_rand[i]))
#adicionar print de quais valores sao maiores que 10 ao final do código
print(f'número de elementos maiores que 10: {contador}')
print(f'valores maiores que 10: {lista}')
```
