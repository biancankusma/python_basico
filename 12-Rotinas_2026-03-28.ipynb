Instancie um objeto da classe NumPy array, contendo 5 valores quantitativos randômicos normalmente distribuídos, com média 10 e desvio padrão 3.<br><br>
Quantos e quais são os valores superiores a 10?

## 02 – Descreva o racional da sua resposta:

R.: Primeiro, para criar esse objeto é preciso importar a biblioteca numpy: import numpy as np

Depois, o método para obter um array de distribuição normal é: np.random.normal()

Dentro dos parenteses, vão os parametros separados por virgula: media, desvio padrao, numero de valores

Para instanciação do objeto: array_rand=np.random.normal(10,3,5)

Segundo, para responder a pergunta, será criada uma função (contagem) que terá dois parâmetros: o objeto que deve ser um array (ele que será avaliado pela função) e um limite: será um valor float (também poderia ser um inteiro, mas optei por float para não ser muito limitante. Esse valor será o parâmetro de comparação dos valores do objeto na condicional)

Então, dentro da função pode ser usada a mesma estrutura de repetição do exercício anterior, que vai partir de um contador em zero e uma lista vazia 

Ao longo de um range definido pelo numero de elementos do objeto, vai verificar uma condição que se dá pelo if.

Se o valor observado for maior que o limite, será acrescentado esse valor à lista e adicionado 1 ao contador. 

Para adicionar valores à lista pode ser usado o método append() levando em conta que são valores float e a indexação do loop a cada volta

Então, a repetição se dá sucessivamente até o final, quando vai devolver fora do loop a contagem de valores maiores que o limite e a lista com estes valores.

## 04 – Escreva o programa em Python:

```python
import numpy as np
array_rand=np.random.normal(10,3,5)
print(array_rand)

def contagem(obj:np.array, limite:float):
    contador=0
    lista=[]
    for i in range(len(obj)):
        if obj[i] > limite:
            contador+=1
            lista.append(float(obj[i]))
    print(f'número de elementos maiores que 10: {contador}')
    print(f'valores maiores que 10: {lista}')
```

```python
contagem(array_rand,10)
```

## 09 – Desenvolva um programa de computador que retorne a divisão de dois objetos simples. Seu programa deve ser capaz de reconhecer a classe destes objetos e o valor, antes de retornar ao usuário.

```python
def divisao(obj1:int|float,obj2:int|float)->float:
    if type(obj1) not in (int, float) or type(obj2) not in (int, float):
        raise TypeError("Os valores devem ser int ou float")
    
    # Evitar divisão por zero
    if obj2 == 0:
        raise ZeroDivisionError("Divisão por zero não é permitida")
    
    # Mostrando informações antes do retorno
    print(f'obj1: valor={obj1}, tipo={type(obj1).__name__}')
    print(f'obj2: valor={obj2}, tipo={type(obj2).__name__}')
    
    div = obj1 / obj2
    
    return div
```

```python
divisao(27,3)
```

## 10 – Desenvolva um programa de computador que retorne a divisão de dois objetos compostos. Seu programa deve ser capaz de reconhecer a classe destes objetos e o valor, antes de retornar ao usuário.

```python
def divisao_arrays(obj1:np.ndarray, obj2:np.ndarray):
    
    if type(obj1) != np.ndarray or type(obj2) != np.ndarray:
        raise TypeError("Os objetos devem ser arrays numpy")
    
    # Verificar formato
    if obj1.shape != obj2.shape:
        raise ValueError("Os arrays devem ter o mesmo formato")
    
    print(f'obj1: tipo={type(obj1).__name__}, valores=\n{obj1}')
    print(f'obj2: tipo={type(obj2).__name__}, valores=\n{obj2}')
    
    # Verificar divisão por zero
    if np.any(obj2 == 0):
        print("Aviso: há divisão por zero no array!")
    
    resultado = obj1 / obj2
    
    return resultado
```

```python
arr1 = np.random.randint(0, 20,(5, 5))
arr2=np.random.randint(1,20,(5,5))
divisao_arrays(arr1,arr2)
```
